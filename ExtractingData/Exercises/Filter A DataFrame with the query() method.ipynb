```python
import pandas as pd
```

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
```

```python
# show dataframe
df
```

# query()

```python
# select players older than 34
df.query("age>34")

# exercise: write the equivalent boolean slicing
#df[df['age']>34]
```

```python
# select players older than 34 from Italy
df.query("age>34 and nationality=='Italy'")

# exercise: write the equivalent boolean slicing
# df[(df['age']>34) & (df['nationality']=='Italy')]
```

```python
# add a not operator to the first example
df.query("not(age>34)")

# exercise: write the equivalent boolean slicing
# df[~(df['age']>34)]
```

```python
# convert height to meters and select those with height above 1.8
df.query("height_cm/100 > 1.8")

# exercise: write the equivalent boolean slicing
# df[df['height_cm']/100>1.8]
```

```python
# Check data types
df.dtypes

# Convert dob column to datetime
df['dob'] = pd.to_datetime(df['dob'])

# Check again
df.dtypes

# Query people born after 1990
df.query("dob.dt.year > 1990")
```
